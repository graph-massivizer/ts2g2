```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from pyunicorn.timeseries import VisibilityGraph

# 生成一个示例时间序列
time_series = np.sin(np.linspace(0, 4 * np.pi, 20))  # 50个点的正弦波

# 创建 Visibility Graph
vg = VisibilityGraph(time_series)

# 获取邻接矩阵（注意：graph 是一个属性，而不是方法）
adj_matrix = vg.graph

# 创建 NetworkX 图
G = nx.Graph()

# 添加节点
num_nodes = len(time_series)
G.add_nodes_from(range(num_nodes))

# 添加边
for i in range(num_nodes):
    for j in range(i + 1, num_nodes):
        if adj_matrix[i, j]:  # 确保有连接
            G.add_edge(i, j)

# 绘制图
plt.figure(figsize=(8, 6))
nx.draw(G, with_labels=True, node_color="lightblue", edge_color="gray", node_size=500)
plt.title("Visibility Graph from Time Series")
plt.show()
```
